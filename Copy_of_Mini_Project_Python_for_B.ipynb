```python
#Gene expression profiling
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("Gene_expression (1).csv")

print (df)
```

```python
print (df.shape)
```

```python
print (df.columns)
```

```python
df.head()
```

```python
#Average Calculation
```

```python
df["Control_Avg"] = (df["Control_1"] + df["Control_2"])/2

df["Treated_Avg"] = (df["Treated_1"] + df["Treated_2"])/2


print (df)
```

```python
#Difference/Comparison analysis
```

```python
df ["Difference"] = (df["Treated_Avg"] - df["Control_Avg"])

print(df)
```

```python
#Expression analysis
```

```python
df["Regulation"] = df["Difference"].apply(lambda x: "Upregulated" if x > 0 else "Downregulated")

print (df[["Gene","Difference","Regulation"]])
```

```python
#Visualization
```

```python
#Bar plot
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.bar(df["Gene"], df["Difference"])
plt.title("Gene Expression Profiling")
plt.xlabel("Gene")
plt.ylabel("Difference")
plt.title("Gene Expression Differences")
plt.xticks(rotation=45)
plt.show()
```

```python
#Expression comparison plot
```

```python
x = range(len(df["Gene"]))

plt.figure(figsize=(10,5))

plt.bar(x, df["Control_Avg"], width=0.4, label="Control")

plt.bar( [i + 0.4 for i in x],
    df["Treated_Avg"],
    width=0.4,
    label="Treated")

plt.xticks([i + 0.2 for i in x],
    df["Gene"],
    rotation=45)

plt.xlabel("Genes")
plt.ylabel("Average Expression")
plt.title("Control vs Treated Gene Expression")

plt.legend()

plt.show()
```

```python
df.to_csv("Gene_expression_analysis.csv", index=False)
```
